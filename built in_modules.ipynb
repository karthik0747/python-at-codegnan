```python
# who will reach top in less steps
import random
steps = rolls = 0
while True:
    input("to roll dice for that clock enter button")
    dice_val = random.randint(1,6)
    rolls += 1
    print("current dice value:", dice_val)
    if steps + dice_val <= 100:
        steps += dice_val
    if steps == 100:
        print(f"you are reached TOP in {rolls} rolls")
        break
        
    print(f"for {rolls} rolls your at {steps} position")
    
```
